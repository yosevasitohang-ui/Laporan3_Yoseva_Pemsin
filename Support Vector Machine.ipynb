```python
# Yoseva Faurensia Sitohang_202431195

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.svm import SVC
from sklearn.metrics import (accuracy_score, confusion_matrix, classification_report)
```

```python
# Yoseva Faurensia Sitohang_202431195

df = pd.read_csv("data.csv")

print("5 Data Pertama:")
display(df.head())

print("\nInformasi Dataset:")
df.info()
print("\nStatistik Deskriptif:")
display(df.describe())
```

```python
# Yoseva Faurensia Sitohang_202431195

df_clean = df.copy()

kolom_hapus = [kolom for kolom in df_clean.columns
    if kolom.lower() == 'id' or kolom.startswith('Unnamed:') ]

df_clean = df_clean.drop(columns=kolom_hapus)

print("Kolom yang dihapus:", kolom_hapus)
print("\nJumlah Nilai Kosong:")
print(df_clean.isna().sum())

kolom_numerik = df_clean.select_dtypes(include=np.number).columns

for kolom in kolom_numerik:
    df_clean[kolom] = df_clean[kolom].fillna(
        df_clean[kolom].median()
    )

df_clean = df_clean.dropna(subset=['diagnosis'])

print("\nNilai Kosong Setelah Pembersihan:")
print(df_clean.isna().sum())
print("\nUkuran Dataset Setelah Pembersihan:")
print(df_clean.shape)
```

```python
# Yoseva Faurensia Sitohang_202431195

print("Nilai Unik Diagnosis:")
print(df_clean['diagnosis'].unique())
print("\nJumlah Setiap Kelas:")
print(df_clean['diagnosis'].value_counts())

# Arti setiap kelas:
# M = Malignant (kanker ganas)
# B = Benign (tumor jinak)
```

```python
# Yoseva Faurensia Sitohang_202431195

le = LabelEncoder()

df_clean['diagnosis_encoded'] = le.fit_transform(
    df_clean['diagnosis'])

print("Pemetaan Kelas:")
for angka, kelas in enumerate(le.classes_):
    print(kelas, "=", angka)

print("\nHasil Encoding Target:")
display(df_clean[['diagnosis', 'diagnosis_encoded']].head())
```

```python
# Yoseva Faurensia Sitohang_202431195

X = df_clean.drop(
    columns=['diagnosis', 'diagnosis_encoded'])

y = df_clean['diagnosis_encoded']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0, stratify=y)

print("Jumlah Data Latih:", len(X_train))
print("Jumlah Data Uji:", len(X_test))
print("\nUkuran X_train:", X_train.shape)
print("Ukuran X_test:", X_test.shape)
print("\nProporsi Kelas Data Latih:")
print(y_train.value_counts(normalize=True))
print("\nProporsi Kelas Data Uji:")
print(y_test.value_counts(normalize=True))
```

```python
# Yoseva Faurensia Sitohang_202431195

pipeline_svm = Pipeline([
    ('scaler', StandardScaler()),
    ('svc', SVC()) ])

print(pipeline_svm)
```

```python
# Yoseva Faurensia Sitohang_202431195

param_grid = {
    'svc__kernel': ['linear', 'rbf', 'poly'],
    'svc__C': [0.1, 1, 10, 100],
    'svc__gamma': ['scale', 0.01, 0.1, 1] }

jumlah_kombinasi = (
    len(param_grid['svc__kernel']) *
    len(param_grid['svc__C']) *
    len(param_grid['svc__gamma']) )

print("Jumlah Kombinasi Parameter:", jumlah_kombinasi)
```

```python
# Yoseva Faurensia Sitohang_202431195

grid_search = GridSearchCV(estimator=pipeline_svm, param_grid=param_grid, cv=5, scoring='accuracy', n_jobs=-1)
grid_search.fit(X_train, y_train)

print("Parameter Terbaik:")
print(grid_search.best_params_)
print("\nSkor Validasi Silang Terbaik:")
print(grid_search.best_score_)
```

```python
# Yoseva Faurensia Sitohang_202431195

hasil_cv = pd.DataFrame(grid_search.cv_results_)

kandidat_terbaik = hasil_cv[
    hasil_cv['rank_test_score'] == 1]

print("Kandidat dengan Peringkat Pertama:")
display(
    kandidat_terbaik[
        ['params', 'mean_test_score', 'std_test_score', 'rank_test_score'] ] )

print("\nJumlah Kandidat Terbaik:")
print(len(kandidat_terbaik))

model_svm = grid_search.best_estimator_
print("\nModel Akhir:")
print(model_svm)
```

```python
# Yoseva Faurensia Sitohang_202431195

y_train_pred = model_svm.predict(X_train)
y_test_pred = model_svm.predict(X_test)

akurasi_train = accuracy_score(y_train, y_train_pred)
akurasi_test = accuracy_score(y_test, y_test_pred)

print("Accuracy Data Latih:", akurasi_train)
print("Accuracy Data Uji:", akurasi_test)

print("\nConfusion Matrix Data Uji:")
cm = confusion_matrix(y_test, y_test_pred)
print(cm)

print("\nClassification Report Data Uji:")
print(
    classification_report(
        y_test,
        y_test_pred,
        target_names=le.classes_) )
```

```python
# Yoseva Faurensia Sitohang_202431195

fitur_dua = ['radius_mean', 'texture_mean']
X_train_2d = X_train[fitur_dua]
X_test_2d = X_test[fitur_dua]

C_terbaik = grid_search.best_params_['svc__C']

model_margin = Pipeline([
    ('scaler', StandardScaler()),
    ('svc', SVC(kernel='linear', C=C_terbaik))])

model_margin.fit(X_train_2d, y_train)
svc_linear = model_margin.named_steps['svc']

print("Koefisien (coef_):")
print(svc_linear.coef_)
print("\nIntercept:")
print(svc_linear.intercept_)
print("\nJumlah Support Vector Setiap Kelas:")
print(svc_linear.n_support_)

w = svc_linear.coef_[0]
norm_w = np.linalg.norm(w)
margin = 2 / norm_w

print("\nNorm W:")
print(norm_w)
print("\nLebar Margin:")
print(margin)
```

```python
# Yoseva Faurensia Sitohang_202431195

data_baru = X_test.iloc[[0]]
minimum_train = X_train.min()
maksimum_train = X_train.max()

validasi_rentang = (
    (data_baru >= minimum_train) &
    (data_baru <= maksimum_train) )

print("Data Baru:")
display(data_baru)
print("\nPemeriksaan Rentang Fitur:")
display(validasi_rentang)

print("\nApakah Semua Fitur Berada dalam Rentang Data Latih?")
print(validasi_rentang.all(axis=1).iloc[0])

prediksi_baru = model_svm.predict(data_baru)
hasil_prediksi = le.inverse_transform(prediksi_baru)

print("\nHasil Prediksi:")
print(hasil_prediksi[0])
```
